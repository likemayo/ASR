## The Command

```bash
cd phase_1
python image_curator.py --load-csv your_images.csv
```

Replace `your_images.csv` with the actual filename of your CSV file.

---

## What This Command Does

| Step | Action | Output |
|------|--------|--------|
| 1 | Reads your CSV file | - |
| 2 | Loads all 100 images and metadata | - |
| 3 | Auto-generates 6 questions per image (600 total) | - |
| 4 | Creates gold answer sets (600 total) | - |
| 5 | Validates all data is complete | ✓ or ✗ |
| 6 | Saves metadata.json | `data/metadata.json` |
| 7 | Saves prompts.jsonl | `data/prompts.jsonl` |
| 8 | Saves gold_answers.jsonl | `data/gold_answers.jsonl` |
| 9 | Prints completion status | Status report |

**Result:** 3 production-ready files for Phase 2 ✅

---

## What You Need Before Running

### Prepare These:

✓ **100 images** in `data/images/` folder
  - Format: JPEG or PNG
  - Examples: `image_001.jpg`, `image_002.jpg`, ... `image_100.jpg`

✓ **CSV file** in `data/` folder with these 15 columns:
  1. `image_id` - Unique ID (001, 002, etc.)
  2. `filename` - Image filename (must match actual file)
  3. `scene_type` - indoor/outdoor/activity
  4. `complexity` - 1, 2, or 3
  5. `subjects` - Who/what is in image (use `;` to separate multiple)
  6. `actions` - What's happening (use `;` to separate multiple)
  7. `colors` - Colors present (use `;` to separate multiple)
  8. `setting` - Location description
  9. `structure_what` - What description
  10. `structure_who` - Who description
  11. `structure_where` - Where description
  12. `structure_color` - Color description
  13. `structure_size` - Size description
  14. `structure_mood` - Mood description
  15. `gold_standard` - Ideal response

**Important:** All 15 columns must have values (no empty cells)

## CSV Example

What your CSV should look like:

```
image_id,filename,scene_type,complexity,subjects,actions,colors,setting,structure_what,structure_who,structure_where,structure_color,structure_size,structure_mood,gold_standard
001,image_001.jpg,outdoor,2,child;dog,playing with a ball,green;brown,park,a child and a dog,a boy and his pet,in a sunny park,green grass and trees,medium-sized dog,happy,A child throws a ball and plays with a dog in the park
002,image_002.jpg,indoor,1,child,reading,blue,classroom,a child reading,a focused student,in a classroom,blue sweater,one child,calm,A child reads a book in a classroom
...
100,image_100.jpg,outdoor,3,children;animals,running,red;yellow,field,multiple subjects in motion,excited children,in an open field,bright colors,large animals,energetic,Children play with animals in a colorful field
```

**Key Rules:**
- Use semicolons (`;`) not commas to separate items in list fields
- All rows must have all 15 values (no empty cells)
- Exactly 100 rows (plus header row)
- Save as `.csv` file

## The 3 Output Files

### File 1: `data/metadata.json`
- **Size:** ~10-15 KB
- **Contains:** All image metadata (scene type, complexity, subjects, actions, colors, setting, structure words, gold standard)
- **Format:** Single JSON object with 100 entries
- **Used by:** Phase 2 for training context

### File 2: `data/prompts.jsonl`
- **Size:** ~80-100 KB
- **Contains:** 600 questions (6 per image)
- **Format:** One JSON per line (JSONL format)
- **Used by:** Phase 2 as training questions for reading comprehension

### File 3: `data/gold_answers.jsonl`
- **Size:** ~60-80 KB
- **Contains:** Acceptable answer variations for each question (600 total)
- **Format:** One JSON per line (JSONL format)
- **Used by:** Phase 2 for answer validation and scoring

## Troubleshooting

### Error: "CSV file not found"
- Make sure your CSV is in the `data/` folder
- Check the filename spelling (case-sensitive)
- Make sure file extension is `.csv`

### Error: "Image not found"
- Check that image filenames in your CSV exactly match the actual filenames in `data/images/`
- Include file extensions in CSV (e.g., `.jpg` or `.png`)

### Error: "Validation failed"
- Check that all 15 columns have values (no empty cells)
- Fix the CSV and run the command again

### No output files created
- Check that validation passed (no errors in output)
- Verify all 100 images have `structure_*` columns filled (these generate the questions)
